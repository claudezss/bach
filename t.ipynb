```python
from bach import get_device
from bach.model import MusicTransformer
import torch
model = MusicTransformer()
model.load_state_dict(torch.load("D:\\Dev\\repo\\bach\\data_cache\\artifacts\\transformer\\epoch_9.pt", map_location="cpu")["model_state_dict"])

device = get_device()

model.to(device)
model.eval()

token = encoded

sequence = []

seq_length = 512

if len(token) > 0:

    for i in range(0, len(token) - seq_length, seq_length // 2):
        seq = token[i : i + seq_length]
        if len(seq) == seq_length:
            sequence.append(seq)

    # Add last sequence if it's not too short
    if len(token) % seq_length > seq_length // 2:
        last_seq = token[-seq_length:]
        if len(last_seq) == seq_length:
            sequence.append(last_seq)

input = sequence[0][:512]
```

```python
decode_midi(input, "input.mid")
```

```python
generated_tokens = model.generate(input, max_length=512, temperature=0.7)
```

```python
decode_midi(generated_tokens, "output.mid")
```
